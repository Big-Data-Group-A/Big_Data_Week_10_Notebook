```python
X=df[["quiz_avg","assignments_avg","attendance_pct"]]
Y= df["score"]

X_train, X_test,Y_train,Y_test=train_test_split(
    X,Y, test_size=0.25, random_state=42
)

print("Train size:", len(X_train))
print("Test size:",  len(X_test))
```

```python

# Exercises 2.1
model = LinearRegression()
model.fit(X_train, Y_train)

for name, coef in zip(X.columns, model.coef_):
    print(f"{name}: {coef:.3f}")
print("Intercept:", round(model.intercept_, 3))

# One extra quiz point is worth about (the quiz_avg coefficient) exam points,
# holding assignments and attendance constant.
```

```python
#exercises 2.2
Y_pred= model.predict(X_test)

mae=mean_absolute_error(Y_test,Y_pred)
r2=r2_score(Y_test,Y_pred)

print("MAE:", round(mae,2))
print("R2", round(r2,2))

# MAE 3.33: on average, the model's prediction is about 3 points away from
# a student's real score.
# R2 0.89: the model explains about 89% of the differences between students'
# scores; the remaining 11% comes from things we did not measure.
```

```python
#Exercises 2.3
new_student=pd.DataFrame({
    "quiz_avg":[70],
    "assignments_avg":[75],
    "attendance_pct": [90],
})

print("Predicted score:", round(model.predict(new_student)[0],1))

me = pd.DataFrame({
    "quiz_avg":[65],
    "assignments_avg":[70],
    "attendance_pct":[85],
})
print("My predicted score:", round(model.predict(me)[0],1))

# The new student (quiz 70, assignments 75, attendance 90) is predicted to score
# about 73.7, which is a comfortable pass (50 is the pass mark).
# My own invented numbers (quiz 65, assignments 70, attendance 85) give a
# predicted score of about 68.4.
# The new data must use the same column names as the training features,
# otherwise the model cannot read it.
```

```python
# Exercise 3.1
df["passed"] = (df["score"] >= 50).astype(int)
print(df["passed"].value_counts())

Xc = df[["quiz_avg", "assignments_avg", "attendance_pct"]]
yc = df["passed"]

Xc_train, Xc_test, yc_train, yc_test = train_test_split(
    Xc, yc, test_size=0.25, random_state=42
)

clf = LogisticRegression(max_iter=1000)
clf.fit(Xc_train, yc_train)

print("Train:", len(Xc_train), "Test:", len(Xc_test))
```

```python
# Exercise 3.2
yc_pred = clf.predict(Xc_test)

print("Model accuracy:", round((yc_pred == yc_test).mean(), 3))
print("Lazy baseline (everyone passes):", round(yc_test.mean(), 3))
print(confusion_matrix(yc_test, yc_pred))

# The two accuracies are identical (0.895), but the model is NOT useless.
# The lazy baseline scores 0.895 only because 34 of the 38 test students
# passed, and it catches 0 of the 4 students who failed.
# The confusion matrix shows what accuracy hides: our model correctly
# identifies 3 of the 4 failing students (top-left) and misses only 1 of
# them (top-right, predicted pass but really failed).
# It also wrongly flags 3 passing students as at risk (bottom-left).
# Since the purpose of the model is to catch at-risk students, the
# confusion matrix is the fair judge, not accuracy alone.
```

```python
# Exercise 4.1
y = df["score"]

X1 = df[["quiz_avg"]]                                        
X2 = df[["quiz_avg", "assignments_avg", "attendance_pct"]]   
X3 = pd.get_dummies(                                         
    df[["quiz_avg", "assignments_avg", "attendance_pct", "program"]],
    columns=["program"], drop_first=True
)

results = {}
for label, Xv in [("1. quiz only", X1), ("2. three numeric", X2), ("3. + program", X3)]:
    Xtr, Xte, ytr, yte = train_test_split(Xv, y, test_size=0.25, random_state=42)
    m = LinearRegression().fit(Xtr, ytr)
    results[label] = round(r2_score(yte, m.predict(Xte)), 2)

for k, v in results.items():
    print(f"{k}: R2 = {v}")

# What helped: adding assignments and attendance (R2 rose from about 0.62 to 0.89),
# because they carry extra information about the score.
# What didn't: adding program (R2 stayed about 0.88), because it adds little beyond
# what the other features already say.
# Lesson: more information only helps if it is relevant. More features is not
# automatically better.
```

# Model Card: Student Score Predictor

**1. Purpose.** This model predicts a student's final score (0 to 100) from their
quiz average, assignments average and attendance. It is meant for AUCA lecturers and
advisors, to spot students who may need support early.

**2. Data and features.** Trained on 150 students (112 for training, 38 held back for
testing). Inputs: quiz_avg, assignments_avg, attendance_pct. Target: score.
A linear regression was used.

**3. Performance.** On the 38 unseen students, the MAE is 3.33, so predictions are
off by about 3 points on average. R2 is 0.89, so the model explains about 89% of the
differences between students' scores. One extra quiz point adds about 0.58 points.

**4. Risks.** A wrong prediction could label a capable student as "at risk" (stigma,
lower expectations) or miss a struggling student who then gets no help. The model was
trained on only 150 students and may not fit other cohorts or programs. It ignores
things like illness, family problems and workload.
- Use it as: an early-warning signal that prompts a conversation and support.
- Do NOT use it for: grading, failing students, scholarships, or discipline.
  A person should always make the final decision.
